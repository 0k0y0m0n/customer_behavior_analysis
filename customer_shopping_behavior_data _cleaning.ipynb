```python
import pandas as pd 

# Reading the Dataset
df = pd.read_csv("C:\\Users\\BOBO\\Documents\\analytics projects\\customer_shopping_behavior.csv")
```

```python
# Quick info
df.info()
```

```python
# Checking for missing values
df.isnull().sum()
```

```python
# Changing the formatting of the columns 
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
df = df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
# Replacing the missing values
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# Creating a new column for age group 
labels = ['Young adult', 'Middle age', 'Adult', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
# Creating another column for frequency of purchase
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14, 
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df = df.drop('promo_code_used', axis=1)
```
